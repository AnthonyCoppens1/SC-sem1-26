```python
import requests
import re

url = "https://en.wikipedia.org/wiki/Elephant"
headers = {'User-Agent': 'Anthony'}
response = requests.get(url,headers = headers)

if response.status_code == 200:
    html = response.text
    regex = r"elephant|Elephant"
    matches = re.findall(regex, html)
    print(len(matches))
else:
    print("try again")
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://en.wikipedia.org/wiki/Elephant"
headers = {'User-Agent' : 'Anthony'}
response = requests.get(url, headers = headers)
html = response.text

soup = BeautifulSoup(html)
#print(soup)
#print(soup.prettify())
print(soup.title)
print(soup.title.get_text())
#print(soup.get_text())
print(soup.get_text()[0:1000].split())
print(soup.get_text().count("Elephant"))
print(soup.get_text().count("elephant"))

h2_tags = soup.find("h2")
print(h2_tags)
h2_tags = soup.find_all("h2")
print(h2_tags)

for title in h2_tags:
    print(title.get_text())
    
```

```python
# getting all the languages
import requests
from bs4 import BeautifulSoup

url = "https://en.wikipedia.org/wiki/Moose"
response = requests.get(url, headers=headers)
print(response.status_code)
soup = BeautifulSoup(response.text)

liTags = soup.find_all("li", {'class': "interlanguage-link"})
print(liTags)

for tag in liTags:
    print(tag.a["title"])
```

```python
from bs4 import BeautifulSoup
import requests
import re

url = "https://en.wikipedia.org/wiki/Pizza"
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text)
imgTags = soup.find_all("img")

print(len(imgTags))
src = [img['src'] for img in imgTags]
results = re.findall(r'[\w-]+\.[jpgpn]{3}(?=\/\d{3,}px)', str(src))

for url in results:
    print(url)

```

```python
import requests
from bs4 import BeautifulSoup

url = "https://www.meteoblue.com/nl/weer/webcams/mount-seymour_canada_6082657"
response = requests.get(url, headers=headers)
soup = BeautifulSoup(response.text)

images = soup.find_all("a", {'class': "webcam-img"})

counter = 0

for image in images:
    url = image.img["src"]
    filename = f"Webcam/image_{counter}.jpg"
    r = requests.get(url, allow_redirects=True)

    with open(filename, 'wb') as file:
        file.write(r.content)
    counter += 1
```
