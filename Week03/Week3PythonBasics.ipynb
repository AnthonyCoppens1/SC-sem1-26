### creating arrays

```python
import numpy as np

list = [1, 5, 8, 10, 3]
print(list)

array = np.array(list)
print(array)

print(list * 2)
print(array * 2)

list.append(2)
#array.add(2) --> not possible

# 1D arrays
print(np.arange(10)) # fill an array with the number from 0 up until the given values (not included)
print(np.zeros(15))# fill an array with x number of zeros
print(np.ones(10)) # fill an array with x number of ones
print(np.full(10, 23)) #fill an array of x size with y number

print(np.random.randint(-5, 6, 10)) #array of random ints between lower and top values, top values not included
print(np.random.random(15)) #15 random floats between 0 and 1
```

## Reshaping and converting and array

```python
import numpy as np
array = np.random.randint(-5, 6, 10)
print(array)
print(array.shape)
print(array.ndim)
print(array.dtype)

#conversion
arrayOfFloats = array.astype('D')
print(arrayOfFloats)
# i - integer / b - boolean / f - float / M - DateTime / S - string / can also work with str, bool, int, ...

arrayofStrings = array.astype(str)
print(arrayofStrings)

array2D = np.reshape(array, (5,2))
print(array2D)
array2D = np.reshape(array, (2,5))
print(array2D)
array2D = np.reshape(array, (10,1))
print(array2D)

#array2D = np.reshape(array, (3,3))
print(array2D)
```

## Slicing arrays
[start : end : step]
- start is the starting index
- end is the ending index not included
- step is the stepsize

```python
import numpy as np
array = np.array([6, 8, 2, 9, 3, 1, 4, 7, 0, 5])
print(array)
print(array[1:5:1])
print(array[1:5:])
print(array[1:5:2])
print(array[0:5:2])
print(array[:5:2])

print(array[::3])
print(array[5::])

print(array[::-1])
print(array[-9:-1:])

print(array[0:10:-1])
print(array[5::-1])
print(array[::-2])
```

### exercise on slicing
generate a list of 24 temperatures, one for every hour of the day.
Isolate the specific parts of the day and analyse them with slicing.

-Make a array of the 24 tempartures (random ints)
-slice these parts
* morning rush: first 6 hours of the day
* late evening: last 4  hours
* trend: every 3 hours -> 00.00, 03.00, 06.00
* lunchbreak: 12.00 - 18.00 (REVERSED)

```python
import random
import numpy as np

temps = [random.randint(-1,26) for x in range(24)]
print(temps)

measurements = np.random.randint(-1,26,24)
print(measurements)

morning = measurements[:6]
print(morning)

lateEvening = measurements[-4::]
print(lateEvening)

trend = measurements[::3]
print(trend)

lunch = measurements[18:11:-1]
print(lunch)
```

## Working with arrays

```python
import numpy as np
a = np.arange(0,10)
b = np.arange(9, -1, -1)
print(a, b)

print(a + b)
print(a - b)
print(a * b)

print(np.where(a % 2 == 0, a + 1, a))
print(np.where(b > 5))
print(np.where(b > 5, b, 0))

b = np.reshape(b, (2,5))
print(b)
print(b.sum(axis= 0)) #all columns
print(b.sum(axis= 1)) #all rows
print()

print(a.sum())
print(a.min())
print(a.max())
print(np.sqrt(a)) #square root of each individual number
print(np.sqrt(a.max()))
```
