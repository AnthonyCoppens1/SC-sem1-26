## Input and output
Conversion between different types
- string --> str / literal text ""
- int --> whole number (integer)
- char --> chr / ' '
- float --> default for decimal numbers
- bool --> true or false

```python
x = input("Give me a number: ")
print(x)
print("Value of x:", x)
print("Value of x: " + x)
print(type(x))

#this is pure text, not taken in account
#conversion

xAsInteger = int(x)
print("Value of x:", xAsInteger)
print(type(xAsInteger))

#tech below is possible, but a lot of back and forth with conversions
#way more convenient to just use the , as a split between datatypes
print("Value of x:" + str(xAsInteger))

#Line above is an example of how to convert back to a string

lilFloatie = float(xAsInteger)
print("Value of x:", lilFloatie)
print(type(lilFloatie))
```

## Combining outputs --> concatenation
2 big variants
- '+' --> combines two datatypes or especially two strings, keep in mind you have to provide a space.
- ',' --> combines two datatypes but provides its own space to split two or more given datatypes

```python
number = 4
print("The number is", number)
print(number + number) #8

print("The number is " + str(number)) #string 
print(str(number) + str(number)) #44
print(str(number + number))
```

## Expressions
### arithmetic operations
operaters: + - * / % // **


```python
#looking at the value of x and maybe y
x, y, z = 10, 3, "test"

print(x + y) #13
print(x - y) #7
print(x * y) #30
print(x / y) #3.333333334
print(x // y) #3
print(x % y) #1
print(x ** y) #1000
```

### Assignment operators

```python
x, y = 10, 3
x += y; print(x) #13
x -= y; print(x) #10
x *= y; print(x) #30
x /= y; print(x) #10.0
x //= y; print(x) #3.0
x = 10

x **= y; print(x) #1000
x %= y; print(x) #1

```

### Comparison operators

```python
x, y = 8, 2
print(x < y) #false
print(x > y) #true
print(x <= y) #false
print(x >= y) #true

print(x == y) #false

x, y = 5, 5.0
print(x == y) #true
print(x != y) #false
```

## Conditional statements
if-structure
Keep in mind zero is an even number!!

```python
number = int(input())
print(number)
if number > 0:
    print("BIGGER")
elif number < 0:
    print("SMALLER")
else:
    print("0")


#detection of even or odd
print("number", number)
if number % 2 == 0:
    print("EVEN")
else:
    print("ODD")


#combining conditions
# and / or / not
number < 0 and number > -10
number % 2 == 0 or number < 50
not(number % 2 == 0)
```
