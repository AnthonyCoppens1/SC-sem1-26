## Input and output
Conversion between different types
- string --> str / literal text ""
- int --> whole number (integer)
- char --> chr / ' '
- float --> default for decimal numbers
- bool --> true or false

```python
x = input("Give me a number: ")
print(x)
print("Value of x:", x)
print("Value of x: " + x)
print(type(x))

#this is pure text, not taken in account
#conversion

xAsInteger = int(x)
print("Value of x:", xAsInteger)
print(type(xAsInteger))

#tech below is possible, but a lot of back and forth with conversions
#way more convenient to just use the , as a split between datatypes
print("Value of x:" + str(xAsInteger))

#Line above is an example of how to convert back to a string

lilFloatie = float(xAsInteger)
print("Value of x:", lilFloatie)
print(type(lilFloatie))
```

## Combining outputs --> concatenation
2 big variants
- '+' --> combines two datatypes or especially two strings, keep in mind you have to provide a space.
- ',' --> combines two datatypes but provides its own space to split two or more given datatypes

```python
number = 4
print("The number is", number)
print(number + number) #8

print("The number is " + str(number)) #string 
print(str(number) + str(number)) #44
print(str(number + number))
```

### Assignment operators

```python
x, y = 10, 3
x += y; print(x) #13
x -= y; print(x) #10
x *= y; print(x) #30
x /= y; print(x) #10.0
x //= y; print(x) #3.0
x = 10

x **= y; print(x) #1000
x %= y; print(x) #1

```

### Comparison operators

```python
x, y = 8, 2
print(x < y) #false
print(x > y) #true
print(x <= y) #false
print(x >= y) #true

print(x == y) #false

x, y = 5, 5.0
print(x == y) #true
print(x != y) #false
```

## Conditional statements
if-structure
Keep in mind zero is an even number!!

```python
number = int(input())
print(number)
if number > 0:
    print("BIGGER")
elif number < 0:
    print("SMALLER")
else:
    print("0")


#detection of even or odd
print("number", number)
if number % 2 == 0:
    print("EVEN")
else:
    print("ODD")


#combining conditions
# and / or / not
number < 0 and number > -10
number % 2 == 0 or number < 50
not(number % 2 == 0)
```

## Loops
- for loop: iterating over a sequence or a range, you know exactly how many repetitions you need to loop.
- while: conditional loop, as long as a given condition is true (or false) the loop will continue.

```python
#FOR LOOP

for c in "a word":
    print(c)

for c in "a word":
    print(c,end = "+")

print()
#check ranges
for x in range(0,9): #upper boundary not included
    print(x)

for x in range(-5, 5):
    print(x, end = " ")

for x in range(0, 11):
    print("5 *", x, "is", x * 5)
    print(f"5 * {x} is {x * 5}")
```

```python
#WHILE LOOPS
value = input()

while value != "STOP":
    print(value)
    value = input()


x = 5
while x <= 10000:
    x *= 5
    print(x, end= " ")


print()

x = 5
while x <= 10000:
    print(x, end = " ")
    x *= 5

```

## Combining condition
- and --> combining conditions
- or --> one or the other or one of the options
- not --> opposite of a given option

```python
x, y, z = 1, 2, 3

print(x < y or y > z) #true
print(x < y and y < z) #true
print(not(x > y or y > z)) #true
```

### exercise on both loops and if-statements
FIZZBUZZ
Write a program that prints out the number 1 - 100. 
- For multiples of 3 print "FIZZ" instead of the number
- For multiples of 5 print "BUZZ" instead of the number
- For a multiple of 3 AND 5, print "FIZZBUZZ" instead of the number

```python
for x in range(1, 101):
    if x % 3 == 0 and x % 5 == 0:
        print("FIZZBUZZ")
    elif x % 3 == 0:
        print("FIZZ")
    elif x % 5 == 0:
        print("BUZZ")
    else:
        print(x)
```
