## recap exercise
write a function that check for a correct email.
rules: 
- a-z0-9_.- are all allowed before @
- no 2x @
- after @ max 2 . but with text in between
- has to end in country code or .com

```python
import re

def validate_email(email):
    if "@" not in email:
        return False

    parts = email.split("@")
    if len(parts) != 2:
        return False

    #back half
    regex = r"[a-z]{2,}(\.[a-z]{2,})?\.[a-z]{2,3}$"
    if re.match(regex, parts[1]) is None:
        return False

    regex = r"\b[\w\.\-]{2,}"
    if re.match(regex, parts[0]) is None:
        return False

    return True

print(validate_email("fatima.@gmail.com"))
print(validate_email("$LamarMoneyDrip$$Let'sGOOOOO@@Lamar.LA"))
print(validate_email("fatima@g.g.com"))
print(validate_email("fatima@ga.gl.com"))
print(validate_email("fatima@gg..com"))
print(validate_email("fatima@help.testing"))
```
