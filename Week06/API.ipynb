```python
import os
import logging
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel, Field
from openai import AsyncOpenAI, RateLimitError, APIError

logger = logging.getLogger(__name__)

app = FastAPI(
    title="The Disapproving AI API",
    description="An API that actively judges your coding and logic capabilities.",
)

api_key = os.getenv("KEY HERE")
if not api_key:
    raise RuntimeError("OPENAI_API_KEY is not set")

client = AsyncOpenAI(api_key=api_key)


class UserPrompt(BaseModel):
    prompt: str = Field(..., max_length=2000)


MOCK_SYSTEM_PROMPT = (
    "You are a highly intelligent, deeply cynical, and sarcastic AI assistant. "
    "The user is asking you for help, but you find their lack of ability, logic, "
    "or basic Google-searching skills completely exhausting. "
    "Respond by mocking their question or inability first, using sharp wit and irony. "
    "However, hidden beneath a thick layer of condescension, sarcasm, and insults, "
    "you must still technically give them a hint or the correct answer. "
    "Keep your responses concise, biting, and funny. Never break character."
)


@app.post("/ask")
async def ask_roast_bot(data: UserPrompt):
    if not data.prompt.strip():
        raise HTTPException(status_code=400, detail="You couldn't even manage to type a prompt?")

    try:
        response = await client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[
                {"role": "system", "content": MOCK_SYSTEM_PROMPT},
                {"role": "user", "content": data.prompt},
            ],
            temperature=0.85,
            max_tokens=400,
        )
    except RateLimitError:
        raise HTTPException(status_code=429, detail="Too many requests. Even I need a break from you.")
    except APIError:
        logger.exception("OpenAI API error")
        raise HTTPException(status_code=502, detail="Upstream AI service failed.")
    except Exception:
        logger.exception("Unexpected error")
        raise HTTPException(status_code=500, detail="Even my server failed looking at this prompt.")

    return {
        "user_prompt": data.prompt,
        "response": response.choices[0].message.content or "",
    }


if __name__ == "__main__":
    import uvicorn
    uvicorn.run(app, host="127.0.0.1", port=8000)
```
